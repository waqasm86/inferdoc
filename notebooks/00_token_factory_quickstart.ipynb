# Token Factory quickstart

Nebius Token Factory serves NVIDIA Nemotron-3 Nano. InferDoc makes the request; Python records the returned text, usage, model, and finish reason. Run JupyterLab from the repository root with InferDoc installed.

## Environment and safe configuration

`InferDocSettings.from_env()` may load a local `.env`; this notebook never displays or stores its secret values. Live calls require both `INFERDOC_RUN_LIVE_TESTS=1` and `NEBIUS_API_KEY`.

```python
import asyncio
import os
import inferdoc
from inferdoc.config import InferDocSettings

settings = InferDocSettings.from_env()
live_enabled = os.getenv("INFERDOC_RUN_LIVE_TESTS") == "1" and bool(settings.api_key)
print({"model": settings.default_model, "endpoint": settings.base_url, "live_enabled": live_enabled})
```

## One short live call

The synchronous `inferdoc.chat` SDK function runs in a worker thread so this cell works in Jupyter’s active event loop. Thinking is disabled for normal workload inference; the Doctor uses Nemotron reasoning separately.

```python
response = None
if live_enabled:
    response = await asyncio.to_thread(
        inferdoc.chat,
        model="nvidia/NVIDIA-Nemotron-3-Nano-30B-A3B",
        prompt="Define TTFT in one sentence.",
        max_tokens=32,
        temperature=0.0,
        chat_template_kwargs={"enable_thinking": False},
    )
    print({"text": response.text, "model": response.model,
           "token_usage": response.usage.model_dump() if response.usage else None,
           "finish_reason": response.finish_reason})
else:
    print("Live call skipped. Set INFERDOC_RUN_LIVE_TESTS=1 and NEBIUS_API_KEY to opt in.")
```
